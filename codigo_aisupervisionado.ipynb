```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
dados = {
    "PM2.5": [10, 15, 45, 60],
    "PM10": [20, 30, 70, 90],
    "Resultado": [0, 0, 1, 1]
}

df = pd.DataFrame(dados)
display(df)
```

```python
X = df[["PM2.5", "PM10"]]
y = df["Resultado"]
```

```python
modelo = DecisionTreeClassifier(random_state=42)
modelo.fit(X, y)

print("Modelo treinado com sucesso!")
```

```python
novo_horario = pd.DataFrame({
    "PM2.5": [35],
    "PM10": [55]
})

previsao = modelo.predict(novo_horario)

print("Previsão:", previsao)

if previsao[0] == 0:
    print("Resultado: Qualidade do ar adequada")
else:
    print("Resultado: Qualidade do ar inadequada")
```

```python
plt.figure(figsize=(12, 7))

plot_tree(
    modelo,
    feature_names=["PM2.5", "PM10"],
    class_names=["Adequada", "Inadequada"],
    filled=True,
    rounded=True
)

plt.title("Árvore de Decisão - Qualidade do Ar")
plt.show()
```

```python
novos_dados = pd.DataFrame({
    "PM2.5": [12, 25, 50, 70],
    "PM10": [25, 45, 75, 100]
})

novas_previsoes = modelo.predict(novos_dados)

resultado_final = novos_dados.copy()
resultado_final["Previsão"] = novas_previsoes
resultado_final["Classificação"] = resultado_final["Previsão"].map({
    0: "Adequada",
    1: "Inadequada"
})

display(resultado_final)
```
